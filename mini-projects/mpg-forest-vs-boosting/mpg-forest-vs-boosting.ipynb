```python
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.model_selection import cross_val_score

mpg = sns.load_dataset('mpg')
mpg = mpg.dropna()
x = mpg[['cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year']]
y = mpg['mpg']

forest = RandomForestRegressor(n_estimators=100, random_state=42)
boost = HistGradientBoostingRegressor(max_iter=100, learning_rate=0.1, random_state=42)

forest_scores = cross_val_score(forest, x, y, cv=5, scoring='neg_mean_absolute_error')
boost_scores = cross_val_score(boost, x, y, cv=5, scoring='neg_mean_absolute_error')

print("Forest folds:", (-forest_scores).round(2))
print("Boost  folds:", (-boost_scores).round(2))
print("Forest mean:", round(-forest_scores.mean(), 2))
print("Boost  mean:", round(-boost_scores.mean(), 2))
```
