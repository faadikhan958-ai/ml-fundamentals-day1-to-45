```python
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
df=sns.load_dataset('flights')
passengers=df['passengers']
lag1=[]
lag12=[]
target=[]
for i in range (12,len(passengers)):
    lag1.append(passengers[i-1])
    lag12.append(passengers[i-12])
    target.append(passengers[i])
data=pd.DataFrame(
    {
        'lag1':lag1,      #pichly month ka data
        'lag12':lag12,    #last year
        'target':target   #real value  
    }
)    
print(data.head())
```

```python
cut=int(len(data)*0.8)
train=data.iloc[:cut]
test=data.iloc[cut:]
model=LinearRegression()
model.fit(train[['lag1','lag12']],train['target'])
pred=model.predict(test[['lag1','lag12']])
print('mae lag1+lag12',round(mean_absolute_error(test['target'],pred),2))
copy_month=abs(test['target']-test['lag1'])
copy_year=abs(test['target']-test['lag12'])
print('copy_month:',round(copy_month.mean(),2))
print('copy_year:',round(copy_year.mean(),2))
```

```python
from sklearn.ensemble import RandomForestRegressor
forest=RandomForestRegressor(n_estimators=100,random_state=42)
forest.fit(train[['lag1','lag12']],train['target'])
pred=forest.predict(test[['lag1','lag12']])
print('mean_absolute_error:',round(mean_absolute_error(test['target'],pred),1))
```
